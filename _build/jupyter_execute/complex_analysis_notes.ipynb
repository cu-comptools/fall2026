## 6.1 Motivation

In smooth quadrature, we saw that many of the error bounds depended on some high derivative of the function to be integrated. For example, 
1. the (simple, not composite) trapezoidal rule had 
$$ \left|\int_a^b f(x) \mathrm{d} x - \frac{b-a}{2}[f(a) + f(b)]\right| \leq \frac{1}{12} ||f''||_{L_{\infty}}(b-a)^3. $$
2. Gauss--Legendre quadrature has
$$\left|\int_a^b f(x) \mathrm{d} x - \frac{b-a}{2}[f(a) + f(b)]\right| \sim \frac{f^{(2n+1)}}{(2n+1)!}.$$

The right-hand-sides above are reminiscent of terms in a Taylor expansion, e.g. consider the Taylor expansion of $f(z)$ around $z = 0$, 
$$
f(z) = \sum_{n=0}^{\infty} a_n z^n,
$$
$$
a_n = \frac{f^{(n)}(0)}{n!}.
$$


## Analyticity

```{admonition} Definition: analyticity
If $f$ is analytic at $0$, there exists $\xi>0$ such that the above series converges absolutely for all $z$ in the disc
$$
|z|<\xi,
$$
and diverges absolutely for all $z$ outside.
```

Analyticity necessarily implies smoothness, i.e. the function has derivatives of all orders, but not the other way round.

Examples:
- polynomials, $e^z$, and trigonometric functions are analytic everywhere ("entire"),
- $|z|$ is not analytic at $z=0$,
- piecewise functions are not analytic where intervals meet.


## Absolute convergence

A real or complex series
$$
\sum_{n=0}^{\infty} a_n
$$
converges **absolutely** if
$$
\sum_{n=0}^{\infty} |a_n| = L
$$
for some real and finite $L$.


## More examples

Consider
$$
f(z) = \frac{1}{1+z^2}.
$$

It has singularities at
$$
z=\pm i,
$$
so there is no convergent Taylor series around those points.

For the Taylor series about $z=0$,
$$
\text{radius of convergence } \xi = 1.
$$

The singularities lie at $+i$ and $-i$ on the imaginary axis.

Within the radius of convergence, take
$$
|z| = \xi-\varepsilon,
\qquad \forall\,\varepsilon>0.
$$

As $n\to\infty$, the terms decrease, so
$$
|a_n|(\xi-\varepsilon)^n \leq C.
$$

That is,
$$
|a_n| \leq \frac{C}{(\xi-\varepsilon)^n}.
$$


Assume $f$ is analytic in an open domain containing the "stadium"
$$
\left\{z\in\mathbb{C}: \operatorname{dist}(z,[a,b])\leq R\right\}.
$$

Then the Taylor coefficients satisfy
$$
|a_n| \leq \frac{C}{R^n}.
$$


## KL divergence

For probability distributions $p$ and $q$,
$$
D_{\mathrm{KL}}(P\,\|\,Q)
= \int_{-\infty}^{\infty}
p(x)\log\!\left(\frac{p(x)}{q(x)}\right)\,dx.
$$
